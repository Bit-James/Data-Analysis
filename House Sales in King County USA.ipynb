# Model Development

Fitting a linear regression model using the  longitude feature `'long'` and  caculating the R^2.

```python
X = df[['long']]
Y = df['price']
lm = LinearRegression()
lm.fit(X,Y)
lm.score(X, Y)
```

Fitting a linear regression model using the feature `'sqft_living'` and calculating the R^2.

```python
U = df[['sqft_living']]
V = df['price']
lm.fit(U,V)
lm.score(U,V)
```

Fitting a linear regression model using the following list of features:

```python
features =["floors", "waterfront","lat" ,"bedrooms" ,"sqft_basement" ,"view" ,"bathrooms","sqft_living15","sqft_above","grade","sqft_living"]    
```

```python
X = df[features]
Y = df['price']
lm.fit(X,Y)
lm.score(X,Y)
```

```python
Input=[('scale', StandardScaler()),('polynomial', PolynomialFeatures(include_bias=False)),('model', LinearRegression())]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.pipeline import FeatureUnion
from sklearn.metrics import mean_squared_error, r2_score
```

```python
pipe = Pipeline(Input)
pipe
```

```python
X = X.astype(float)
pipe.fit(X,Y)
ypipe=pipe.predict(X)
print(r2_score(Y,ypipe))
```

Based on the model development steps, which include more relevant features and employing more sophisticated modeling techniques, significantly improves the predictive power of the model for estimating 'prices'. The increasing R² scores evidence this information:

- 0.000468 when using only the 'long' attribute.
- 0.493 when using only the 'sqft_living' attribute.
- 0.658 when using all other attributes in a linear regression.
- 0.751 when using a pipeline that incorporates StandardScaler, PolynomialFeatures, and LinearRegression with all attributes

### Model Evaluation and Refinement¶

The data is splitted into training and testing sets:

```python
features =["floors", "waterfront","lat" ,"bedrooms" ,"sqft_basement" ,"view" ,"bathrooms","sqft_living15","sqft_above","grade","sqft_living"]    
X = df[features]
Y = df['price']

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.15, random_state=1)


print("number of test samples:", x_test.shape[0])
print("number of training samples:",x_train.shape[0])
```

```python
RigeModel=Ridge(alpha=0.1)
RigeModel.fit(x_train, y_train)
RigeModel.score(x_test, y_test)
```

```python
pr = PolynomialFeatures(degree=2)
x_train_pr = pr.fit_transform(x_train)
x_test_pr = pr.fit_transform(x_test)

RigeModel=Ridge(alpha=0.1)
RigeModel.fit(x_train_pr, y_train)
RigeModel.score(x_test_pr, y_test)
```

## conclusion

The R² score for the testing data improved significantly after applying a polynomial transformation of degree 2 to the features.
- Initial Model: The R² score for the Ridge regression model without polynomial features was approximately 0.65.
- Polynomial Transformed Model: After transforming the features with PolynomialFeatures(degree=2), the R² score for the testing subset increased to approximately 0.70.

This indicates that the relationship between the features and the target variable is better captured by a non-linear model, and the polynomial transformation helped improve the model's performance on unseen data.
